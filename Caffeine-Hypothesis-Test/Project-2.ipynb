# Caffeine Content: Illustrative One-Sample t-Test

This notebook demonstrates a two-sided one-sample **t-test** against an illustrative benchmark of 120 mg. The 50 hard-coded values have no documented external provenance in this repository, so they are treated as illustrative/synthetic-or-unsourced data rather than verified product measurements. Likewise, 120 mg is treated as an illustrative benchmark, not as a sourced company claim.

See `DATA_PROVENANCE.md` for the provenance status, assumptions, and requirements for empirical use.

## Statistical question

For this demonstration:

- $H_0: \mu = 120$ mg
- $H_A: \mu \ne 120$ mg
- Significance level: $\alpha = 0.05$

A one-sample t-test is used because the repository does not justify a known population standard deviation. Variability is therefore estimated from the sample.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import t as student_t
```

```python
# Illustrative caffeine-content values in milligrams.
# Provenance is not established in this repository; do not treat these as verified measurements.
data = np.array([
    118, 122, 125, 117, 119, 121, 123, 116, 124, 120,
    118, 121, 126, 115, 117, 123, 122, 119, 124, 118,
    116, 120, 123, 127, 121, 122, 119, 124, 125, 118,
    117, 123, 121, 120, 116, 122, 123, 125, 118, 121,
    119, 120, 117, 124, 122, 118, 121, 123, 120, 116
])

mu0 = 120      # illustrative benchmark mean, not a sourced company claim
alpha = 0.05   # significance level
```

```python
n = len(data)
df = n - 1
sample_mean = data.mean()
sample_sd = data.std(ddof=1)
standard_error = sample_sd / np.sqrt(n)
t_stat = (sample_mean - mu0) / standard_error
critical_value = student_t.ppf(1 - alpha / 2, df=df)
p_value = 2 * student_t.sf(abs(t_stat), df=df)
ci_low = sample_mean - critical_value * standard_error
ci_high = sample_mean + critical_value * standard_error
decision = 'Reject H0' if p_value < alpha else 'Fail to Reject H0'

print(f'Sample size: {n}')
print(f'Sample mean: {sample_mean:.4f} mg')
print(f'Sample standard deviation: {sample_sd:.4f} mg')
print(f'Standard error: {standard_error:.4f} mg')
print(f'Degrees of freedom: {df}')
print(f't-statistic: {t_stat:.4f}')
print(f'Critical t-value: ±{critical_value:.4f}')
print(f'P-value: {p_value:.4f}')
print(f'95% confidence interval: [{ci_low:.4f}, {ci_high:.4f}] mg')
print(f'Decision: {decision}')
```

For the embedded illustrative values, the expected results are approximately: sample mean 120.5800 mg, sample standard deviation 2.9972 mg, $t(49)=1.3683$, two-sided $p=0.1774$, and a 95% confidence interval of [119.7282, 121.4318] mg.

```python
plt.figure(figsize=(8, 6))
plt.hist(data, bins=10, edgecolor='black')
plt.axvline(sample_mean, linestyle='--', linewidth=2, label=f'Sample mean = {sample_mean:.2f} mg')
plt.axvline(mu0, linestyle=':', linewidth=2, label=f'Illustrative benchmark = {mu0} mg')
plt.title('Illustrative Caffeine-Content Values')
plt.xlabel('Caffeine content (mg)')
plt.ylabel('Frequency')
plt.legend()
plt.show()
```

## Assumptions and limitations

A standard one-sample t-test is normally interpreted under a sampling process in which observations are independent and the mean is adequately modeled by the t procedure. Because the source and collection process for these values are not documented, independence and external validity cannot be verified.

The analysis is therefore a statistical demonstration only. Reproducibility of the calculation does not establish provenance of the data. A real product or company analysis would require a documented dataset, sampling or measurement procedure, a sourced benchmark or label claim, and a review of the test assumptions.

## Conclusion

At $\alpha=0.05$, the illustrative sample fails to reject the null hypothesis that the mean equals the 120 mg benchmark. The 95% confidence interval also contains 120 mg.

This result does **not** verify a company's caffeine-content claim and does not prove that the true mean is exactly 120 mg. It only shows that, for these illustrative values and this model, the observed difference from 120 mg is not statistically significant at the 5% level.
